## Feature engineering

- YearRemodAdd. Remodel date
- YearBuilt. Original construction date


Calculate the columns and the current year

```python
train_df = data.copy()
```

```python
train_df['is_train'] = 1
test_df['is_train'] = 0
```

```python
processed_df = pd.concat([train_df, test_df], ignore_index=True)
```

```python
import datetime
```

```python
current_year = datetime.datetime.now().year
```

```python
processed_df['YearRemodAdd']
```

```python
processed_df['YearBuilt']
```

```python
processed_df['age_since_built'] = current_year - processed_df['YearBuilt']
processed_df['years_since_remod'] = current_year - processed_df['YearRemodAdd']
processed_df['is_remodeled'] = np.where((processed_df['YearRemodAdd'] - processed_df['YearBuilt']) > 0, 1, 0)
```

### NA values

```python
processed_df.isna().sum()[processed_df.isna().sum() > 0]
```

```python
processed_df['Alley'].notna().sum()
```

```python
processed_df['Alley'] = processed_df['Alley'].fillna('None')
```

```python
LotFrontage_median = processed_df['LotFrontage'].median()
```

```python
processed_df['LotFrontage'] = processed_df['LotFrontage'].fillna(LotFrontage_median)
```

```python
processed_df['MasVnrType'] = processed_df['MasVnrType'].fillna('None')
```

```python
processed_df['MasVnrArea']= processed_df['MasVnrArea'].fillna(0.0)
```

```python
processed_df['BsmtQual']= processed_df['BsmtQual'].fillna('None')
```

```python
processed_df['BsmtCond']= processed_df['BsmtCond'].fillna('None')
```

```python
processed_df['BsmtExposure']= processed_df['BsmtExposure'].fillna('None')
```

```python
processed_df['BsmtFinType1']= processed_df['BsmtFinType1'].fillna('None')
```

```python
processed_df['BsmtFinType2']= processed_df['BsmtFinType2'].fillna('None')
```

```python
processed_df['Electrical']= processed_df['Electrical'].fillna('None')
```

```python
processed_df['FireplaceQu']= processed_df['FireplaceQu'].fillna('None')
```

```python
processed_df['GarageType'] = processed_df['GarageType'].fillna('None')
```

```python

processed_df['GarageYrBlt'] = processed_df['GarageYrBlt'].fillna(processed_df['YearBuilt'])
```

```python
processed_df['GarageFinish'] = processed_df['GarageFinish'].fillna('None')
```

```python
processed_df['GarageQual'] = processed_df['GarageQual'].fillna('None')
```

```python
processed_df['GarageCond'] = processed_df['GarageCond'].fillna('None')
```

```python
processed_df['PoolQC'] = processed_df['PoolQC'].fillna('None')
```

```python
processed_df['Fence'] = processed_df['Fence'].fillna('None')
```

```python
processed_df['MiscFeature'] = processed_df['MiscFeature'].fillna('None')
```

```python
processed_df.columns
```

```python
processed_df.drop(['YearBuilt','YearRemodAdd','Id'],axis=1, inplace=True)
```

```python
processed_df.select_dtypes(exclude=['number']).columns.tolist()
```

### Trasnforming quality features

```python
quality_map = {'None':0, 'Po':1, 'Fa':2, 'TA':3, 'Gd':4, 'Ex':5}
quality_cols = ['ExterQual', 'ExterCond', 'BsmtQual', 'BsmtCond', 'HeatingQC', 'KitchenQual', 'FireplaceQu', 'GarageQual', 'GarageCond', 'PoolQC']

for col in quality_cols:
    processed_df[col] = processed_df[col].map(quality_map)
```

```python
processed_df['LotShape'] = processed_df['LotShape'].map({'IR3':0, 'IR2':1,'IR1':2, 'Reg':3})
```

```python
processed_df['Utilities'] = processed_df['Utilities'].map({'AllPub':3, 'NoSewr':2,'NoSeWa':1,'ELO':0})
```

```python
processed_df['LandSlope'] = processed_df['LandSlope'].map({'Gtl':2, 'Mod':1, 'Sev':0})
```

```python
processed_df['BsmtExposure'] = processed_df['BsmtExposure'].map({'Gd':4, 'Av':3, 'Mn':2, 'No':1, 'NA':0})
```

```python
map_bsmt_fin = {'None': 0, 'Unf': 1, 'LwQ': 2, 'Rec': 3, 'BLQ': 4, 'ALQ': 5, 'GLQ': 6}
processed_df['BsmtFinType1'] = processed_df['BsmtFinType1'].map(map_bsmt_fin)
processed_df['BsmtFinType2'] = processed_df['BsmtFinType2'].map(map_bsmt_fin)
```

```python
processed_df['GarageFinish'] = processed_df['GarageFinish'].map({'Fin':3, 'RFn':2, 'Unf':1, 'NA':0})
```

```python
processed_df['PavedDrive'] = processed_df['PavedDrive'].map({'Y':2, 'P':1, 'N':0})
```

```python
processed_df['Fence'] = processed_df['Fence'].map({'NA':0, 'MnWw':1, 'GdWo':2,'MnPrv':3,'GdPrv':4})
```

```python
map_func = {'Sal': 0, 'Sev': 1, 'Maj2': 2, 'Maj1': 3, 'Mod': 4, 'Min2': 5, 'Min1': 6, 'Typ': 7}
processed_df['Functional'] = processed_df['Functional'].map(map_func)
```

```python
processed_df['CentralAir'] =  processed_df['CentralAir'].map({'N':0, 'Y':1})
```

```python
processed_df = pd.get_dummies(processed_df, drop_first=True, dtype=int)
```

```python
categorical_columns= processed_df.select_dtypes(exclude=['number']).columns.tolist()
print(f"Categorical columns missing to transform: {len(categorical_columns)}")
```

```python
train_final = processed_df[processed_df['is_train'] == 1].copy()
X_train = train_final.drop(columns=['is_train', 'SalePrice'])
y_train = train_final['SalePrice']

# Separar el set de Prueba (Test) para la predicción final
test_final = processed_df[processed_df['is_train'] == 0].copy()
X_test = test_final.drop(columns=['is_train', 'SalePrice'])

print(f"Forma de X_train: {X_train.shape}")
print(f"Forma de X_test: {X_test.shape}")
```

```python
X_train['Fence'] = X_train['Fence'].fillna(0)
X_test['Fence'] = X_test['Fence'].fillna(0)
```

```python
X_train['BsmtExposure'] = X_train['BsmtExposure'].fillna(0)
X_test['BsmtExposure'] = X_test['BsmtExposure'].fillna(0)

X_train['GarageFinish'] = X_train['GarageFinish'].fillna(0)
X_test['GarageFinish'] = X_test['GarageFinish'].fillna(0)
```

```python
print("NA values in X_train:", X_train.isna().sum().sum())
```

```python
X_train.isna().sum()[X_train.isna().sum() > 0]
```

```python
cols_num_test = [
    'BsmtFinSF1', 'BsmtFinSF2', 'BsmtUnfSF', 'TotalBsmtSF', 
    'BsmtFullBath', 'BsmtHalfBath', 'GarageCars', 'GarageArea'
]
X_test[cols_num_test] = X_test[cols_num_test].fillna(0)
```

```python
cols_esenciales_test = ['Utilities', 'KitchenQual', 'Functional']
for col in cols_esenciales_test:
    moda_train = X_train[col].mode()[0] 
    X_test[col] = X_test[col].fillna(moda_train)
```

```python
print("NA values in X_train:", X_train.isna().sum().sum())
print("NA in X_test:", X_test.isna().sum().sum())
```

```python
print("Categorical columns:", X_train.select_dtypes(exclude=['number']).columns.tolist())


print("¿X_train y X_test have the same number of columns:", X_train.shape[1] == X_test.shape[1])
```

## XGBoost implementation

```python
import xgboost as xgb
from sklearn.metrics import root_mean_squared_error
from sklearn.model_selection import train_test_split
```

```python
xgb_model = xgb.XGBRegressor(
    n_estimators = 500,
    max_depth = 5,
    learning_rate=0.05,
    random_state=42,
    n_jobs=-1  
)
```

```python
X_local_train, X_local_val, y_local_train, y_local_val = train_test_split(X_train, y_train, test_size=0.2, random_state=42)
```

```python
xgb_model.fit(
    X_local_train,
    y_local_train,
    eval_set=[(X_local_val, y_local_val)],
    verbose=100, 
)
```

```python
preds_val = xgb_model.predict(X_local_val)
rmse_local = root_mean_squared_error(y_local_val, preds_val)
```

```python
print(f"RMSE Local: {rmse_local:.2f}")
```

Test

```python
y_local_train_log = np.log1p(y_local_train)
y_local_val_log = np.log1p(y_local_val)


xgb_model.fit(
    X_local_train, y_local_train_log,
    eval_set=[(X_local_val, y_local_val_log)],
    verbose=False
)
```

```python

preds_val_log = xgb_model.predict(X_local_val)


preds_val_reales = np.expm1(preds_val_log)


nuevo_rmse = root_mean_squared_error(y_local_val, preds_val_reales)
print(f"RMSE Local logar: {nuevo_rmse:.2f}")
```

```python
rmse_log_oficial = root_mean_squared_error(y_local_val_log, preds_val_log)
print(f"Tu puntuación oficial aproximada (Log-RMSE): {rmse_log_oficial:.4f}")
```
